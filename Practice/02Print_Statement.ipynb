#02. Print Statement in Python
Introduction

The print() statement is used to display output on the screen. It is one of the first and most commonly used functions in Python.

We can use print() to display text, numbers, variables and calculated results.

##Printing Text

Text can be displayed by putting it inside quotation marks.

```python
print("Hello, Python!")
print("Welcome to my notebook.")
```

#Printing Numbers

Numbers can be printed without quotation marks.

```python
print(25)
print(150)
print(7.5)
```

#Printing Variables

We can print the value stored inside a variable.

```python
student_name = "Anisha"
student_age = 21

print(student_name)
print(student_age)
```

#Printing Calculated Results

The print() function can also display calculations.

```python
num1 = 18
num2 = 12

print(num1 + num2)
print(num1 * num2)
```

#Displaying Multiple Values

We can print multiple values using commas.

```python
name = "Suman"
age = 23

print("Name:", name, "Age:", age)
```

#Using Separator

The sep parameter is used to change the separator between multiple values.

```python
print("2026", "08", "22", sep="-")
```

```python
print("Apple", "Mango", "Orange", sep=" | ")
```

#Using End

Normally, print() moves to a new line after displaying output.

The end parameter can change this behavior.

```python
print("Hello", end=" ")
print("World")
```
