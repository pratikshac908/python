#12. Lists in Python
##Introduction

A list is a collection used to store multiple values in one variable. Lists are ordered and changeable.

##Creating a List

```python
fruits = ["Apple", "Mango", "Orange", "Banana"]

print(fruits)
```

##Different Data Types in a List

```python
data = ["Riya", 21, 5.4, True]

print(data)
```

##Positive Indexing

```python
fruits = ["Apple", "Mango", "Orange", "Banana"]

print(fruits[0])
print(fruits[2])
```

##Negative Indexing

```python
print(fruits[-1])
print(fruits[-2])
```

##Slicing

```python
print(fruits[1:3])
```

##Changing List Elements

```python
fruits[1] = "Grapes"

print(fruits)
```

##Adding Elements
append()

```python
fruits.append("Kiwi")

print(fruits)
```

##insert()

```python
fruits.insert(1, "Papaya")

print(fruits)
```

##Removing Elements

```python
fruits.remove("Apple")

print(fruits)
```

##Another method:

```python
fruits.pop()

print(fruits)
```

Length of a List

```python
print(len(fruits))
```

##Sorting

```python
numbers = [45, 12, 78, 23, 9]

numbers.sort()

print(numbers)
```

##Reversing

```python
numbers.reverse()

print(numbers)
```

##Searching

```python
students = ["Asha", "Bina", "Kiran"]

print("Bina" in students)
```

##Counting

```python
numbers = [2, 4, 2, 6, 2, 8]

print(numbers.count(2))
```

##Clearing a List

```python
items = ["Pen", "Book", "Bag"]

items.clear()

print(items)
```
