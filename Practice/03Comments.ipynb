#03. Comments in Python
##Introduction

Comments are notes written inside a program to explain the code. Python does not execute comments.

Comments make programs easier to understand and are useful when working on larger programs.

#Single-Line Comments

A single-line comment starts with#
 This is my first Python comment

```python
print("Hello Python")
```

##Comment After Code

A comment can also be written after a statement.

```python
name = "Riya"  # Store student's name

print(name)
```

##Using Comments to Explain Code

```python
# Store two numbers
number1 = 14
number2 = 6

# Add the two numbers
total = number1 + number2

print(total)
```

##Temporarily Disabling Code

Comments can be used when testing a program.

```python
print("First statement")

# print("This statement is temporarily disabled")

print("Last statement")
```

##Multi-Line Explanations

Python does not have a separate official multi-line comment symbol. We can use multiple # lines.

```python
# This program calculates
# the total marks of
# three subjects.

marks1 = 70
marks2 = 80
marks3 = 75

print(marks1 + marks2 + marks3)
```

##Comments vs Normal Text

A comment is ignored by Python.

```python
# This is a comment

message = "This is normal text"

print(message)
```

##Observation

I learned that comments help explain code without affecting the program output. They are also useful when testing code or temporarily disabling a statement.
