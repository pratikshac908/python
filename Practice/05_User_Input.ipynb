#05. User Input in Python
##Introduction

User input allows a program to receive information from the user. Python uses the input() function for this purpose.

##Taking Text Input

```python
name = input("Enter your name: ")

print("Hello", name)
```

##Input is Stored in a Variable

```python
city = input("Enter your city: ")

print("You live in", city)
```

##Default Type of Input

The value received from input() is normally a string.

```python
age = input("Enter your age: ")

print(age)
print(type(age))
```

##Converting Input to Integer

```python
age = int(input("Enter your age: "))

print(age)
print(type(age))
```

##Taking Decimal Input

```python
price = float(input("Enter product price: "))

print(price)
print(type(price))
```

##Addition Using User Input

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

total = num1 + num2

print("Total:", total)
```

##Common Mistakes
Forgetting int() when taking integer input.
Forgetting float() for decimal input.
Entering text where a number is expected.
Using input directly in calculations without conversion.
