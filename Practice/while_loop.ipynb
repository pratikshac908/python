#Print numbers from 1 to 10
###We start from 1 and continue the loop until the number becomes 10.

```python
# Print numbers from 1 to 10

i = 1

while i <= 10:
    print(i)
    i = i + 1
```

#Print numbers from 10 to 1
###Were, we start from 10 and decrease the value by 1 in every loop.

```python
i = 10

while i >= 1:
    print(i)
    i = i - 1
```

#Print even numbers up to 50
###We start from 2 and increase the number by 2 each time, so only even numbers are printed.


```python
i = 2

while i <= 50:
    print(i)
    i = i + 2
```

#Print the sum of the first 10 numbers
###We use a total variable to add numbers from 1 to 10.

```python
# Find the sum of first 10 numbers

i = 1
total = 0

while i <= 10:
    total = total + i
    i = i + 1

print(total)
```

#Print the multiplication table of 3
###We multiply 3 by numbers from 1 to 10.

```python
i = 1

while i <= 10:
    print(3 * i)
    i = i + 1
```

#Count the digits in a number
###We repeatedly divide the number by 10. Each division removes one digit.

```python
number = 12345
count = 0

while number > 0:
    number = number // 10
    count = count + 1

print(count)
```

#Reverse a number
###We take the last digit of the number and add it to the reversed number.

```python
number = 1234
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 10

print(reverse)
```

#Print squares of numbers from 1 to 10
###We start from 1 and print the square of each number until 10.

```python
i = 1

while i <= 10:
    print(i ** 2)
    i = i + 1
```

#Print all digits of a number
###We get each digit using % 10 and then remove the last digit using // 10.

```python
number = 12345

while number > 0:
    digit = number % 10
    print(digit)
    number = number // 10
```

#Print the factorial of a number
###Factorial means multiplying a number by all positive numbers before it. For example, 5! = 5 × 4 × 3 × 2 × 1.

```python
# Find the factorial of a number

number = 5
factorial = 1
i = 1

while i <= number:
    factorial = factorial * i
    i = i + 1

print(factorial)
```
