#06. Strings in Python
##Introduction

A string is a sequence of characters used to represent text. Strings can contain letters, numbers, spaces and special characters.

##Creating Strings

```python
name = "Sanjay"
city = 'Dharan'

print(name)
print(city)
```

##String Length

The len() function gives the number of characters in a string.

```python
word = "Computer"

print(len(word))
```

##Positive Indexing

String indexing starts from 0.

```python
word = "Python"

print(word[0])
print(word[2])
print(word[5])
```

##Negative Indexing

Negative indexing starts from the end.

```python
word = "Python"

print(word[-1])
print(word[-2])
print(word[-6])
```

##String Slicing

Slicing is used to get a part of a string.

```python
text = "Programming"

print(text[0:4])
print(text[4:8])
```

##Useful String Methods
Uppercase

```python
name = "pratiksha"
print(name.upper())
```

##Lowercase

```python
name = "PRATIKSHA"
print(name.lower())
```

##Replace

```python
text = "I like Java"
print(text.replace("Java", "Python"))
```

##Strip

```python
text = "   Hello   "
print(text.strip())
```

##Checking String Content

```python
sentence = "I am learning Python"

print("Python" in sentence)
print("Java" in sentence)
```

##Strings are Immutable

Strings cannot be changed directly after creation.

```python
word = "Hello"

# word[0] = "Y"   # This causes an error
```

##Instead, we create a new string.

```python
word = "Hello"
word = "Y" + word[1:]

print(word)
```
