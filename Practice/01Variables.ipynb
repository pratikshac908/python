# 01. Variables in Python
## Introduction

A variable is a name used to store a value in Python.
Variables are useful because they allow us to store information
and use that information later in a program.

For example, we can store a person's name, age, city or marks
inside variables.

```python
name = "Pratiksha"
age = 22
city = "Biratnagar"

print(name)
print(age)
print(city)
```

## Observation

Here, name, age, and city are variables. They store different
values, and we can use the variables whenever we need those values.

## Creating and Assigning Variables

In Python, we create a variable by assigning a value using the =
operator.

The variable name is written on the left side and the value is
written on the right side.

```python
student_name = "Ram"
student_age = 20
marks = 85

print(student_name)
print(student_age)
print(marks)
```

## Variable Naming Rules

The following rules should be followed when naming variables:

- A variable name can contain letters, numbers and underscores.
- A variable name cannot start with a number.
- Spaces are not allowed in variable names.
- Python variable names are case-sensitive.
- Python keywords cannot be used as variable names.
- Meaningful variable names are better than unclear names.

```python
student_name = "Sita"
student_age = 21
total_marks = 450

print(student_name)
print(student_age)
print(total_marks)
```

## Valid and Invalid Variable Names

Examples of valid variable names:

- name
- student_name
- 2age
- total_mark

Examples of invalid variable names:

- 2age because it starts with a number.
- student_name because it contains a space.
- student_name because - is not allowed in a normal variable name.

```python
name = "Pratiksha"
age2 = 22
student_name = "Ram"

print(name)
print(age2)
print(student_name)
```

## Case Sensitivity

Python is case-sensitive. This means that uppercase and lowercase
letters are treated as different characters.

For example, name and Name are two different variables.

```python
name = "Ram"
Name = "Sita"

print(name)
print(Name)
```

## Observation

The output is different because name and Name are treated as
different variables in Python.

## Storing Different Types of Values

A variable can store different types of values such as text,
integers, decimal numbers and Boolean values.

```python
name = "Pratiksha"
age = 22
height = 5.4
is_student = True

print(name)
print(age)
print(height)
print(is_student)
```

## Multiple Assignment

Python allows us to assign values to multiple variables in one line.

```python
print(name)
print(age)
print(city)
```

```python
x = y = z = 10

print(x)
print(y)
print(z)
```

## Observation

Multiple variables can receive values in one statement.
This makes the code shorter and easier to write.

## Changing Variable Values

The value stored in a variable can be changed later by assigning
a new value to the same variable.

```python
age = 20

print(age)

age = 22

print(age)
```

## Checking the Type of a Variable

The type() function is used to check the data type of a value
stored in a variable.

```python
name = "Pratiksha"
age = 22
height = 5.4
is_student = True

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

## Common Mistakes

Some common mistakes when working with variables are:

1. Starting a variable name with a number.
2. Using spaces in variable names.
3. Forgetting quotation marks around text.
4. Using the wrong spelling of a variable name.
5. Forgetting that Python is case-sensitive.
6. Using Python keywords as variable names.
