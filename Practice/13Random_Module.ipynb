#13. Random Module
##Introduction

A module is a file containing Python code that provides functions and features we can use in our programs.

The random module is used to generate random values and make random selections.

##Importing Random Module

```python
import random
```

##Random Decimal Number

random.random() generates a random decimal number between 0 and 1.

```python
import random

number = random.random()

print(number)
```

##Random Integer

randint() generates a random integer between the given numbers.

```python
import random

number = random.randint(1, 50)

print(number)
```

##Random Number Within a Range

```python
number = random.randrange(10, 51, 5)

print(number)
```

###This selects a number from the given range using the specified step.

##Selecting a Random Item

```python
import random

fruits = ["Apple", "Mango", "Orange", "Banana"]

choice = random.choice(fruits)

print(choice)
```

##Selecting Multiple Items

sample() selects multiple unique items.

```python
students = ["Asha", "Rina", "Kiran", "Suman", "Bikash"]

selected = random.sample(students, 2)

print(selected)
```

##Shuffling a List

shuffle() changes the order of items randomly.

```python
numbers = [1, 2, 3, 4, 5]

random.shuffle(numbers)

print(numbers)
```

```python
import random

dice = random.randint(1, 6)

print("Dice result:", dice)
```

##Simple Random Game

```python
import random

secret_number = random.randint(1, 10)

guess = int(input("Guess a number between 1 and 10: "))

print("Secret number:", secret_number)

if guess == secret_number:
    print("Correct guess!")
else:
    print("Try again!")
```
