#10. Operators in Python
##Introduction

Operators are symbols used to perform operations on values and variables. Python has different categories of operators.

Screenshot 2026-08-24 223256.png

```python
a = 20
b = 6

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

##Assignment Operators

Assignment operators are used to assign and update values.Screenshot 2026-08-24 223833.png

##Comparison Operators

Comparison operators return True or False.
image.png

```python
a = 15
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

##Logical Operators

Logical operators are and, or and not.

```python
age = 22
has_id = True

print(age > 18 and has_id)
print(age > 25 or has_id)
print(not has_id)
```

##Identity Operators

is and is not check whether two variables refer to the same object.

```python
a = [1, 2]
b = a
c = [1, 2]

print(a is b)
print(a is c)
```

##Membership Operators

in and not in check whether a value exists in a collection.

```python
fruits = ["Apple", "Mango", "Orange"]

print("Mango" in fruits)
print("Grapes" not in fruits)
```

##Operators with Strings

```python
first = "Good"
second = "Morning"

print(first + " " + second)
print(first * 2)
```
