#11. Operator Precedence
##Introduction

Operator precedence determines the order in which Python evaluates different operators in an expression.

Parentheses have the highest priority and can be used to change the normal order.

```python
result = 10 + 5 * 2

print(result)
```

##Using Parentheses

```python
result = (10 + 5) * 2

print(result)
```

###Here, addition happens first because of parentheses.

##Comparison and Logical Operators

```python
age = 22

result = age > 18 and age < 30

print(result)
```

##More Examples

```python
print(20 - 4 * 3)
print((20 - 4) * 3)
```

##Arithmetic Precedence

A simple order to remember is:

Parentheses ()

Exponentiation **

Multiplication, division, floor division and modulus

Addition and subtraction

Comparison operators

Logical not

Logical and

Calculate without parentheses.

```python
print(8 + 4 * 3)
```

Use parentheses.

```python
print((8 + 4) * 3)
```

Compare both expressions.

```python
print(15 + 5 * 2)
print((15 + 5) * 2)
```

Use division and addition.

```python
print(30 + 12 / 3)
```

Use parentheses.

```python
print((30 + 12) / 3)
```

Use comparison and logical operators.

```python
x = 15

print(x > 10 and x < 20)
```

Create your own precedence example.

```python
result = (12 + 8) * 2 - 5

print(result)
```

##Overall Observation

I learned that Python follows a specific order when evaluating expressions. Parentheses can be used when we want a particular operation to happen first.
