#07. String Concatenation
##Introduction

String concatenation means joining two or more strings together.

##Joining Two Strings

```python
first_name = "Aarav"
last_name = "Sharma"

full_name = first_name + " " + last_name

print(full_name)
```

##Combining Text and Variables

```python
name = "Mina"
city = "Pokhara"

print("My name is " + name + " and I live in " + city)
```

##Problem with Different Data Types

We cannot directly concatenate a string and an integer using +.

```python
age = 21

# print("My age is " + age)
```

##Using Type Conversion

```python
age = 21

print("My age is " + str(age))
```

##Using f-Strings

f-strings are a simple modern way of formatting output.

```python
name = "Kusum"
age = 20

print(f"My name is {name} and I am {age} years old.")
```

##Using Multiple Values

```python
name = "Ramesh"
marks = 87

print("Name:", name, "Marks:", marks)
```
