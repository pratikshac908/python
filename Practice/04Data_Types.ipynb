#04. Data Types in Python
##Introduction

Data types tell us what kind of value is stored in a variable. Python has different data types for numbers, text, Boolean values and collections.

##Integer
int is used for whole numbers.

```python
age = 24
students = 35

print(age)
print(students)
print(type(age))
```

##String

str is used for text.

```python
name = "Kabita"
country = "Nepal"

print(name)
print(type(name))
```

##Boolean

Boolean values can be True or False.

```python
is_logged_in = True
is_raining = False

print(is_logged_in)
print(is_raining)
print(type(is_logged_in))
```

##List

A list stores multiple values.

```python
fruits = ["Apple", "Banana", "Mango"]

print(fruits)
print(type(fruits))
```

##Tuple

A tuple is another collection type.

```python
days = ("Sunday", "Monday", "Tuesday")

print(days)
print(type(days))
```

##Set

A set stores unique values.

```python
numbers = {2, 4, 6, 8}

print(numbers)
print(type(numbers))
```

##Dictionary

A dictionary stores data in key-value pairs.

```python
student = {
    "name": "Nisha",
    "age": 20
}

print(student)
print(type(student))
```

##None

None represents the absence of a value.

```python
result = None

print(result)
print(type(result))
```

##Same Operation with Different Data Types

The + operator behaves differently depending on the data type.

```python
print(10 + 5)
print("10" + "5")
```

##Checking Data Types

```python
a = 50
b = 12.5
c = "Python"
d = False

print(type(a))
print(type(b))
print(type(c))
print(type(d))
```
