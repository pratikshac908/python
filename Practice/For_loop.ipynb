#Print numbers from 1 to 10
###We use a for loop to print numbers starting from 1 and ending at 10.


```python
# Print numbers from 1 to 10

for i in range(1, 11):
    print(i)
```

#Print all even numbers from 1 to 20
###An even number is a number that can be divided by 2 without a remainder. We use % to check the remainder.

```python
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

#Print all odd numbers from 1 to 20
###An odd number gives a remainder when divided by 2 We check this using %

```python
for i in range(1, 21):
    if i % 2 != 0:
        print(i)
```

#Print all elements of a list
###We create a list of fruits and use a for loop to print each item one by one.

```python
fruits = ["pineapple", "Mango", "watermelon", "papaya"]

for fruit in fruits:
    print(fruit)
```

#Print each character of a string
###A string contains characters. The for loop can take each character one by one and print it.

```python
name = "Pratiksha"

for char in name:
    print(char)
```

#Print the multiplication table of 5
###We use a for loop from 1 to 10 and multiply each number by 5.

```python
# Multiplication table of 5

for i in range(1, 11):
    print(5 * i)
```

#Print the squares of numbers from 1 to 10
###The square of a number means multiplying the number by itself. In Python, we can use ** 2.

```python
for i in range(1, 11):
    print(i ** 2)
```

#Print numbers in reverse order from 10 to 1
###Here, we start from 10 and decrease the number by 1 each time. So we use -1 as the step.

```python
for i in range(10, 0, -1):
    print(i)
```

#Print all vowels in a string
###The vowels are a, e, i, o, u. We check each character of the string and print it if it is a vowel.

```python
text = "I am learning Python"

for char in text:
    if char.lower() in "aeiou":
        print(char)
```
