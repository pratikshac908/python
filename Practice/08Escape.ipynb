#08. Escape Characters
##Introduction

Escape characters are special characters used inside strings. They start with a backslash \.

##New Line

\n creates a new line.

```python
print("Hello\nWorld")
```

##Tab

\t creates a tab space.

```python
print("Name:\tSita")
print("Age:\t21")
```

##Double Quotation Mark

We can use \" to display quotation marks.

```python
print("She said, \"Python is easy.\"")
```

##Single Quotation Mark

```python
print('It\'s a beautiful day.')
```

##Backslash

\\ is used to display a backslash.

```python
print("C:\\Users\\Student")
```

##Combining Escape Characters

```python
print("Name:\tRiya\nCourse:\tPython\nLevel:\tBeginner")
```
